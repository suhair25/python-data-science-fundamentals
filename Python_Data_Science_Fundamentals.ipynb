```python

# Python & Data Science Fundamentals
# Project: Student Performance Analysis

# 1. Variables and Lists
students = ["Dilsath", "Manaseer", "Ilma", "Suhair", "Hassan"]
marks = [78, 85, 45, 92, 35]

print("STUDENT PERFORMANCE ANALYSIS")
print("-" * 35)

# 2. Dictionary
student_data = {
    students[i]: marks[i]
    for i in range(len(students))
}

print("\nStudent Marks:")
for name, mark in student_data.items():
    print(name, ":", mark)

# 3. Function
def calculate_average(scores):
    return sum(scores) / len(scores)

average = calculate_average(marks)
print("\nAverage Mark:", round(average, 2))

# 4. Highest and Lowest Marks
print("Highest Mark:", max(marks))
print("Lowest Mark:", min(marks))

# 5. Pass / Fail using conditions and loops
print("\nPass / Fail Results:")
for name, mark in student_data.items():
    if mark >= 50:
        print(name, "- Pass")
    else:
        print(name, "- Fail")

# 6. Count passed and failed students
passed = sum(1 for mark in marks if mark >= 50)
failed = len(marks) - passed

print("\nPassed Students:", passed)
print("Failed Students:", failed)

# 7. File Handling
with open("student_report.txt", "w") as file:
    file.write("STUDENT PERFORMANCE REPORT\n")
    file.write(f"Average Mark: {average:.2f}\n")
    file.write(f"Highest Mark: {max(marks)}\n")
    file.write(f"Lowest Mark: {min(marks)}\n")
    file.write(f"Passed: {passed}\n")
    file.write(f"Failed: {failed}\n")

print("\nReport saved to student_report.txt")
```

```python

import matplotlib.pyplot as plt

plt.bar(students, marks)
plt.title("Student Performance Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.show()
```
